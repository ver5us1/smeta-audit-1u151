# Проверка проектно-сметной документации

Пошаговый анализ данных ВКР по объекту 1У-151. Все суммы — тыс. руб.

```python
import sys
from pathlib import Path

import pandas as pd

ROOT = Path.cwd()
if not (ROOT / "data").exists():
    ROOT = Path("../")
sys.path.insert(0, str(ROOT / "src"))
from smeta_audit.analysis import (
    aggregate_detailed_cases,
    aggregate_estimated_impacts,
    build_kpi_table,
    load_processed_data,
)
from smeta_audit.pipeline import reproduce_excel_calculations

data = load_processed_data(ROOT / "data/processed")
```

## 1. Структура замечаний

```python
remarks = data["remarks"].sort_values("share_pct", ascending=False)
remarks
```

## 2. Детальный и оценочный расчёт

```python
kpi = build_kpi_table(data["detailed"], data["estimated"])
kpi
```

```python
detail = aggregate_detailed_cases(data["detailed"])
estimated = aggregate_estimated_impacts(data["estimated"])
print(f"Подтверждённый чистый итог: {detail['published_net_conservative']:.2f} тыс. руб.")
print(
    f"Расширенный итог: {detail['published_net_conservative'] + estimated['published_net']:.2f} тыс. руб."
)
```

## 3. Воспроизведение Excel-формул
